```python
import time
import random

# الگوریتم O(n): جمع عناصر لیست
def sum_linear(arr):
    return sum(arr)

# الگوریتم O(n²): پیدا کردن جفت‌هایی که مجموعشان برابر عدد خاصی باشد
def find_pairs_quadratic(arr, target):
    pairs = []
    n = len(arr)
    for i in range(n):
        for j in range(i+1, n):
            if arr[i] + arr[j] == target:
                pairs.append((arr[i], arr[j]))
    return pairs

# اندازه‌گیری زمان
sizes = [1000, 5000, 10000, 20000]
results = []

for size in sizes:
    data = [random.randint(1, 100) for _ in range(size)]

    # تست O(n)
    start = time.time()
    sum_linear(data)
    t_linear = time.time() - start

    # تست O(n²) — فقط برای اندازه‌های کوچکتر
    if size <= 5000:
        start = time.time()
        find_pairs_quadratic(data, 50)
        t_quad = time.time() - start
    else:
        t_quad = "Too slow"

    results.append((size, t_linear, t_quad))

print("Size\tO(n)\t\tO(n²)")
for r in results:
    print(f"{r[0]}\t{r[1]:.6f}\t{r[2]}")
```
